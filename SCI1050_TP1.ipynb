- **`Job Title`** : Le titre du poste occupé par l'employé au sein de l'entreprise.
- **`Hire Date`** : La date d'embauche de l'employé, également au format `YYYY-MM-DD`.
- **`Salary`** : Le salaire annuel de l'employé, exprimé en dollars.
- **`Department`** : Le département de l'entreprise auquel l'employé est rattaché, parmi cinq départements possibles : Ressources Humaines (HR), Informatique (IT), Finance, Marketing, et Opérations.
