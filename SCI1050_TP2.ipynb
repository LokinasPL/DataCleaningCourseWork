L'objectif est d'extraire et structurer les informations essentielles en utilisant le web scraping. Cela inclut l'envoi de requêtes HTTP, l'analyse du HTML pour récupérer les données clés (titre, auteur, prix, disponibilité, etc.), ainsi que la gestion des cas particuliers (erreurs, données manquantes). ## Question 1

À partir du site **Books to Scrape** ([https://books.toscrape.com/](https://books.toscrape.com/)), écrire un code permettant d'extraire les informations suivantes pour chaque livre :  

- **UPC** : Identifiant unique du livre.  
- **Title** : Titre du livre.  
- **Category** : Catégorie du livre (ex. Travel, Mystery, Historical Fiction, etc.).  
- **Price Net** : Prix hors taxes (`Price (excl. tax)`), excluant la devise (£), ex. `51.77` pour `£51.77`.  
- **Price Gross** : Prix avec taxes (`Price (incl. tax)`), excluant la devise (£).  
- **Tax** : Montant de la taxe (`price_gross = price_net + tax`), excluant la devise (£).  
- **Rating** : Évaluation du livre en nombre d'étoiles (de 0 à 5).  
- **Number of Reviews** : Nombre de votes (`Number of reviews`).  
- **Availability** : Nombre de livres disponibles en stock, extrait sous forme d'entier.  
- **Description** : Texte présent dans la section **"Product Description"**.  
- **Detail Link** : URL menant à la page détaillée de chaque livre (ex. `https://books.toscrape.com/catalogue/a-light-in-the-attic_1000/index.html`).  


**Résultat attendu**  

- **Format des résultats** :  
  - Les données extraites doivent être **converties en un DataFrame Pandas** pour une manipulation et une analyse facilitées.  
  - Chaque ligne du DataFrame doit correspondre à un **livre unique**, avec ses informations structurées en colonnes.  

- **Respect des noms de colonnes** :  
  - Les noms de colonnes doivent **strictement respecter** la nomenclature suivante :  

| **Nom de la colonne**  | Type | **Description** |
|------------------------|--|--------------|
| `"upc"`               | `string` |Identifiant unique du livre |
| `"title"`             | `string` |Titre du livre (non tronqué) |
| `"category"`          | `string` |Catégorie du livre (Travel, Mystery, etc.) |
| `"price_net"`         | `float` |Prix hors taxes (float, notation anglaise, ex. `51.77`) |
| `"price_gross"`       | `float` |Prix TTC (float, notation anglaise, ex. `51.77`) |
| `"tax"`               | `float` |Montant de la taxe (`price_gross - price_net`, float) |
| `"rating"`            | `int` |Évaluation du livre en étoiles (de 0 à 5) |
| `"number_of_reviews"` | `int` |Nombre d'évaluations (entier) |
| `"availability"`      | `int` |Nombre d'exemplaires en stock (entier) |
| `"description"`       | `string` |Texte de la description (peut être tronqué) |
| `"detail_link"`       | `string` |Lien vers la page du livre |

**Remarques Importantes pour la collecte des données**  

- **Gestion de la pagination** : Le script doit parcourir **toutes les pages du site** de manière automatique, en suivant les liens de navigation pour collecter l'ensemble des livres disponibles.  

- **Localisation des informations** :  
  - **Sur la page principale** : Certaines données (titre, prix, évaluation, disponibilité, lien vers la page du livre) sont directement accessibles depuis la liste des livres affichés.  
  - **Sur la page détaillée du livre** : D'autres informations (description, UPC, catégorie, prix détaillé, taxes, nombre d'avis) nécessitent une navigation vers la page du livre.  
  - **Informations présentes sur les deux pages** : Certaines données comme l'évaluation peuvent être affichées à la fois sur la page principale et sur la page détaillée, il convient de s'assurer de leur cohérence.  

- **Se concentrer uniquement sur la collecte des données** :  
  - **Si une information est absente**, elle doit être remplacée par `NaN` pour éviter les valeurs erronées.  
  - **Aucune transformation des valeurs** n'est requise à ce stade :  
    - Les prix doivent être **extraits tels qu'affichés** sur le site (sans le symbole de la devise `£` si présent).  
    - L'évaluation (nombre d'étoile) doit être **extraite telle qu'elle**, convertie en valeur numérique.  

- **Respect des formats indiqués** :  
  - **Les prix (`price_net`, `price_gross`, `tax`) doivent être des flottants** (ex. `£51.77` → `51.77`).  
  - **La disponibilité doit être convertie en entier** (ex. `In stock (22 available)` → `22`).
  - **L'évaluation doit être convertie en entier**.  
  - **Le titre ne doit pas être tronqué** pour conserver l'intégralité du nom du livre.  
  - **La description peut être tronquée** si elle est trop longue ou absente sur certaines pages.  



```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd

###Format des informations sur la page du produit
##Tableau d'information
# <table class="table table-striped">
#   <tbody>
#       <tr>
#           <th>UPC</th>
#           <td>a897fe39b1053632</td>
#       </tr>
#       <tr>
#           <th>Product Type</th>
#           <td>Books</td>
#       </tr>
#       <tr>
#           <th>Price (excl. tax)</th>
#           <td>£51.77</td>
#       </tr>
#       <tr>
#           <th>Price (incl. tax)</th>
#           <td>£51.77</td>
#       </tr>
#       <tr>
#           <th>Tax</th>
#           <td>£0.00</td>
#       </tr>
#       <tr>
#           <th>Availability</th>
#           <td>In stock (22 available)</td>
#       </tr>
#       <tr>
#           <th>Number of reviews</th>
#           <td>0</td>
#       </tr>
#   </tbody>
# </table>

##Titre
#    <div class="col-sm-6 product_main">
#       <h1>A Light in the Attic</h1>
#    ...
#    </div>

##Description
#   <div id="product_description" class="sub-header">
#       <h2>Product Description</h2>
#   </div>
#   <p>
#       DESCRIPTION
#   </p>

##Categorie
#   <ul class="breadcrumb">
#       <li>
#         <a href="../../index.html">Home</a>
#       </li>
#       <li>
#         <a href="../category/books_1/index.html">Books</a>
#       </li>
#       <li>
#         <a href="../category/books/poetry_23/index.html">Poetry</a>
#       </li>
#       <li class="active">A Light in the Attic</li>
#   </ul>


#ouvrir page d'accueil
url = 'https://books.toscrape.com/'
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')
products_list = []

#tant qu'il reste des page a lire
has_next = True
while (has_next) :
    #pour chaque livre dans la page
    for book in soup.find_all('article', attrs={'class':'product_pod'}):
      #trouve et monte le lien vers le produit
      product_link = book.find('h3').find('a')['href']
      product_link = urljoin(url, product_link)
      #ouvre la page du produit
      response_product = requests.get(product_link)
      soup_product = BeautifulSoup(response_product.content, 'html.parser')
      ##extrait toutes les informations dans cette page, met NaN si l'information est introuvable ou genere une erreur
      #trouve l'identifiant upc dans le tableau d'information selon le formatage html de la page
      try :
        upc = soup_product.find('table', attrs={'class':'table table-striped'}).find('tr').find('td').text
      except :
        upc = 'NaN'
      #trouve le titre selon le formatage html de la page
      try :
        title = soup_product.find('div', attrs={'class':'col-sm-6 product_main'}).find('h1').text
      except :
        title = 'NaN'
      #trouve la categorie selon le formatage html de la page
      try :
        category = soup_product.find('ul', attrs={'class':'breadcrumb'}).find_all('a')[2].text
      except :
        category = 'NaN'
      #trouve le prix avant taxe dans le tableau d'information selon le formatage html de la page, le convertie en float
      try :
        price_net = soup_product.find('table', attrs={'class':'table table-striped'}).find_all('tr')[2].find('td').text[1:]
        #retire le signe de devise s'il est present
        if price_net[0].isdigit() == False :
          price_net = price_net[1:]
        price_net = float(price_net)
      except :
        price_net = 'NaN'
      #trouve le prix apres taxe dans le tableau d'information selon le formatage html de la page, le convertie en float
      try:
        price_gross = soup_product.find('table', attrs={'class':'table table-striped'}).find_all('tr')[3].find('td').text[1:]
        #retire le signe de devise s'il est present
        if price_gross[0].isdigit() == False :
          price_gross = price_gross[1:]
        price_gross = float(price_gross)
      except :
        price_gross = 'NaN'
      #trouve la taxe dans le tableau d'information selon le formatage html de la page, la convertie en float
      try :
        tax = soup_product.find('table', attrs={'class':'table table-striped'}).find_all('tr')[4].find('td').text[1:]
        #retire le signe de devise s'il est present
        if tax[0].isdigit() == False :
          tax = tax[1:]
        tax = float(tax)
      except :
        tax = 'NaN'
      #trouve la note en string selon le formatage html de la page
      try :
        text_rating = soup_product.find('p', attrs={'class':'star-rating'})['class'][1]
      except :
        text_rating = 'NaN'
      #convertie le chiffre ecris en son equivalent int
      match text_rating:
        case 'One':
          rating = 1
        case 'Two':
          rating = 2
        case 'Three':
          rating = 3
        case 'Four':
          rating = 4
        case 'Five':
          rating = 5
        case _:
          rating = 'NaN'
      #trouve le nombre d'avis dans le tableau d'information selon le formatage html de la page, le convertie en int
      try :
        number_of_reviews = int(soup_product.find('table', attrs={'class':'table table-striped'}).find_all('tr')[6].find('td').text)
      except :
        number_of_reviews = 'NaN'
      #trouve le nombre d'exemplaire disponnible dans le tableau d'information selon le formatage html de la page, le convertie en int, si introuvable on prend pour acquis que 0 sont disponnible
      try :
        availability = int(soup_product.find('table', attrs={'class':'table table-striped'}).find_all('tr')[5].find('td').text.split(' ')[2][1:])
      except :
        availability = 'NaN'
      #trouve la description selon le formatage html de la page
      try :
        description = soup_product.find('div', attrs={'id':'product_description', 'class':'sub-header'}).find_next('p').text
      except :
        description = 'NaN'
      #bati le dictionnaire du produit et l'ajoute a la liste, utilise le product_link utilise pour ouvrir la page pour detail_link
      product_entry = {"upc":upc, "title":title, "category":category, "price_net":price_net, "price_gross":price_gross, "tax":tax, "rating":rating, "number_of_reviews":number_of_reviews, "availability":availability, "description":description, "detail_link":product_link}
      products_list.append(product_entry)
    #check s'il y a une page suivante au catalogue et l'ouvrir
    if soup.find('li', attrs={'class':'next'}):
        url = urljoin(url, soup.find('li', attrs={'class':'next'}).find('a')['href'])
        response = requests.get(url)
        soup = BeautifulSoup(response.content, 'html.parser')
    else:
        has_next = False

#transforme la liste en dataframe manipulable
books = pd.DataFrame(products_list)
print(books.head())
print(books.dtypes)
```

**Explication**
Les étapes du code sont expliquées par les commentaires. On utilise des variables différentes pour la page du livre en court et de la page du catalogue en cours afin de pouvoir les observer en même temps dans le while-loop. Pour chaque information à extraire on utilise un bloc try-except afin de ne pas bloquer le code par une erreur si un des champs n'est pas présent opu est vide dans la page. Si un erreur est lancée, on remplace le champs par NaN tel que prescript par les instructions. Pour traduire la note en int à partir du chiffre écris en lettre, on utilise un match-case pour faire correspondre les chiffres écris aux chiffres en int correspondant, et si on trouve quelque chose à l'extérieur des options attendues entre 1 et 5, on utilse NaN. Pour les prix et les tax, on vérifie si le champs contient la devise et on la retire avant de tansformer le chiffre en float. On a bati une liste vide au début qui allait contenir tous les produits une fois le while-loop terminé, et une fois que toutes les informations sont extraites, on crée un dictionnaire représentant le produit puis on ajoute ce dictionnaire à la-dite liste. Cette liste sera ensuite transformée en Dataframe. On a choisit ce format pusiqu'il est le plus simple à implémenter et manipuler concidérant les boucles utilisées pour l'extraction de données et il permet de facilement assurer les noms et formats des données de toutes les colonnes à travers tout le Dataframe.

```python
import pandas as pd

df_suppl = pd.read_csv('https://raw.githubusercontent.com/hgueziri/MiscData/refs/heads/main/books_suppl.csv')
print(df_suppl.head())
print(df_suppl.isna().sum())
#print(df_suppl.dtypes)
#modification apporter aux types de donnees pour rendre les comparaisons a la question 3 plus facile
for i in df_suppl.index :
  if (not pd.isna(df_suppl.at[i, 'price_net'])) :
    df_suppl.at[i, 'price_net'] = float(df_suppl.at[i, 'price_net'][1:])
  if (not pd.isna(df_suppl.at[i, 'price_gross'])) :
    df_suppl.at[i, 'price_gross'] = float(df_suppl.at[i, 'price_gross'][1:])
  if (not pd.isna(df_suppl.at[i, 'tax'])) :
    df_suppl.at[i, 'tax'] = float(df_suppl.at[i, 'tax'][1:])
  if (not pd.isna(df_suppl.at[i, 'availability'])) :
    df_suppl.at[i, 'availability'] = int(df_suppl.at[i, 'availability'])
```

**Explication**
On utilise read_csv pour lire le fichier csv donné dans les instruction. La fonction head() nous permet de lire les 5 premières rangées du Dataframe, on utilise print() pour les afficher. La fonction isna() extrait toutes les instance de valeur manquante (NaN, Null, etc.) dans chaque colonne, et la fonction sum() permet de faire la somme de ces instance par colonne, puis on utilise print() pour afficher le tableau de résultat.

Ensuite, on passe à travers la base de données pour modifier le type des données des colonnes qui sont sensé être numérique (int ou float) mais qui ne le sont pas dans le fichier original, a moins que la valeur soit NaN. De ce fait, on devient capable de comparer et faire des manipulations mathématiques sur ces valeurs entre les valeurs du df original et ce df_suppl. Pour les prix et tax, on en profite pour retirer la devise (d'où la sélection de toute la chaine sauf le premier charactère à l'indice 0 avant d'être tranformer en float), qui serait un problème de plus a gérer dans la comparaison de ces valeurs autrement.

```python
import pandas as pd
from fuzzywuzzy import fuzz
from numpy import mean

# Fonction pour calculer la similarité entre deux chaînes de caractères
def calculate_similarity(str1, str2):
    if pd.isna(str1) or pd.isna(str2):
        return 0
    else :
      return fuzz.ratio(str1, str2)

# Identifie les doublons selon la similirate dans la colonne nommee
def identify_semantic_duplicates(df, column_name, threshold=75):
    duplicates = []
    for i in range(len(df)):
        for j in range(i + 1, len(df)):
          if isinstance(column_name, str):
            similarity = calculate_similarity(str(df.iloc[i, new_df.columns.get_loc(column_name)]), str(df.iloc[j, new_df.columns.get_loc(column_name)]))
          elif isinstance(column_name, list):
            similarity = mean([calculate_similarity(str(df.iloc[i, new_df.columns.get_loc(col)]), str(df.iloc[j, new_df.columns.get_loc(col)])) for col in column_name])
          else :
            raise ValueError("column_name doit être une chaîne de caractères ou une liste de chaînes de caractères")
          if similarity >= threshold:
              duplicates.append((i, j))
    return duplicates

#Identifie de quel source vient chaque enregistrement
df['origine']='originale'
df_suppl['origine']='supplementaire'

#Concatene les deux dataframes
new_df=pd.concat([df, df_suppl])

#Identifie les index des colonnes pour les manip a venir
iupc = new_df.columns.get_loc('upc')
ititle = new_df.columns.get_loc('title')
icategory = new_df.columns.get_loc('category')
iprice_net = new_df.columns.get_loc('price_net')
iprice_gross = new_df.columns.get_loc('price_gross')
itax = new_df.columns.get_loc('tax')
irating = new_df.columns.get_loc('rating')
inumber_of_reviews = new_df.columns.get_loc('number_of_reviews')
iavailability = new_df.columns.get_loc('availability')
idescription = new_df.columns.get_loc('description')
idetail_link = new_df.columns.get_loc('detail_link')
iorigine = new_df.columns.get_loc('origine')

def gestion_doublons(dataframe, list_doublons) :
  ##Gestion des doublons idententifier dans la liste
  for (i,j) in list_doublons:

    #s'assure qu'on garde et modifie le bon doublon de la bonne source
    if (dataframe.iloc[i, iorigine]=='originale')&(dataframe.iloc[j, iorigine]=='supplementaire') :
      og = i
      supp = j
    elif (dataframe.iloc[j, iorigine]=='originale')&(dataframe.iloc[i, iorigine]=='supplementaire') :
      og = j
      supp = i
    else :
    #si les entrees ne font pas partie des table og et supp, on passe a la prochaine paire de doublons
      continue

    #si les upc sont pas NaN et sont differents, on passe a la prochaine paire, ce ne sont pas des vrais doublons
    if not pd.isna(dataframe.iloc[og, iupc]) and not pd.isna(dataframe.iloc[supp, iupc]) and not dataframe.iloc[og, iupc]==dataframe.iloc[supp, iupc]:
      continue
    #si il manque un upc, lui donner celui de l'autre entree
    elif not pd.isna(dataframe.iloc[og, iupc]):
      dataframe.iloc[supp, iupc] = dataframe.iloc[og, iupc]
    else:
      dataframe.iloc[og, iupc] = dataframe.iloc[supp, iupc]

    #garder le titre de og, sauf si nan, then garder le titre de supp
    if not pd.isna(dataframe.iloc[og, ititle]):
      dataframe.iloc[supp, ititle] = dataframe.iloc[og, ititle]
    else :
      dataframe.iloc[og, ititle] = dataframe.iloc[supp, ititle]

    #garder la categorie de og sauf si nan, then garder la categorie de supp
    if not pd.isna(dataframe.iloc[og, icategory]):
      dataframe.iloc[supp, icategory] = dataframe.iloc[og, icategory]
    else :
      dataframe.iloc[og, icategory] = dataframe.iloc[supp, icategory]

    #garder le prix net de supp sauf si nan, then garder nan
    if not pd.isna(dataframe.iloc[supp, iprice_net]):
      dataframe.iloc[og, iprice_net] = dataframe.iloc[supp, iprice_net]
    else :
      dataframe.iloc[og, iprice_net] = 'NaN'
      dataframe.iloc[supp, iprice_net] = 'NaN'

    #garder le prix de supp sauf si nan, then garder nan
    if not pd.isna(dataframe.iloc[supp, iprice_gross]):
      dataframe.iloc[og, iprice_gross] = dataframe.iloc[supp, iprice_gross]
    else :
      dataframe.iloc[og, iprice_gross] = 'NaN'
      dataframe.iloc[supp, iprice_gross] = 'NaN'

    #garder la tax de supp sauf si nan, then garder nan
    if not pd.isna(dataframe.iloc[supp, itax]):
      dataframe.iloc[og, itax] = dataframe.iloc[supp, itax]
    else :
      dataframe.iloc[og, itax] = 'NaN'
      dataframe.iloc[supp, itax] = 'NaN'

    #garder le rating de og sauf si nan, then garder le rating de supp
    if not pd.isna(dataframe.iloc[og, irating]):
      dataframe.iloc[supp, irating] = dataframe.iloc[og, irating]
    else:
      dataframe.iloc[og, irating] = dataframe.iloc[supp, irating]

    #garder le nombre de review de og sauf si nan, then garder le nombre de review de supp
    if not pd.isna(dataframe.iloc[og, inumber_of_reviews]):
      dataframe.iloc[supp, inumber_of_reviews] = dataframe.iloc[og, inumber_of_reviews]
    else:
      dataframe.iloc[og, inumber_of_reviews] = dataframe.iloc[supp, inumber_of_reviews]

    #garder le nombre de dispo de og sauf si nan, then garder le nombre de dispo de supp
    if not pd.isna(dataframe.iloc[og, iavailability]):
      dataframe.iloc[supp, iavailability] = dataframe.iloc[og, iavailability]
    else:
      dataframe.iloc[og, iavailability] = dataframe.iloc[supp, iavailability]

    #garder la description de og, puisque supp n'en a pas
    dataframe.iloc[supp, idescription] = dataframe.iloc[og, idescription]

    #garder le lien de og, puisque supp n'en a pas
    dataframe.iloc[supp, idetail_link] = dataframe.iloc[og, idetail_link]

  #drop les doublons completement identiques (sauf pour l'index et l'origine), juste en garder un exemplaire
  dataframe.drop_duplicates(subset=['upc', 'title', 'category', 'price_net', 'price_gross', 'tax', 'rating', 'number_of_reviews', 'availability', 'description', 'detail_link'], inplace=True)

  #retourne le dataframe resultant de toutes ces operations
  return dataframe

##Trouver tous les doublons selon les upc, puis les gerer
dub_upc = identify_semantic_duplicates(new_df, 'upc', 100)
new_df = gestion_doublons(new_df, dub_upc)

##Trouver les doublons selon leur titre, puis les gerer
dub_titles = identify_semantic_duplicates(new_df, 'title', 75) #75% de similarite pour gerer les typos
new_df = gestion_doublons(new_df, dub_titles)

##Trouver les doublons selon leur categorie et leur prix gross et leur rating et leur disponnibilite
dub_extra = identify_semantic_duplicates(new_df, ['category', 'price_gross', 'rating', 'availability'], 100)
new_df = gestion_doublons(new_df, dub_extra)

#On retire la colonne origine qui seervait aux manipulations
new_df.drop(columns=['origine'], inplace=True)

#Ordonner les donnees par upc puis imprimer le debut et la fin du dataframe
new_df.sort_values(by='upc', inplace=True)
print(new_df.head())
print(new_df.tail())
```

```python
##On comprend qu'on s'interressse uniquement a la base de donnes df_updated, et qu'on n'a pas a la concatener avec l'ancienne base resultante de la question precedante
import pandas as pd
from fuzzywuzzy import fuzz
from numpy import mean

##On reutilises la fonction pour trouver les doublons implementer dans la question precedante
# Fonction pour calculer la similarité entre deux chaînes de caractères
def calculate_similarity(str1, str2):
    if pd.isna(str1) or pd.isna(str2):
        return 0
    else :
      return fuzz.ratio(str1, str2)

# Identifie les doublons selon la similirate dans la colonne nommee
def identify_semantic_duplicates(df, column_name, threshold=75):
    duplicates = []
    for i in range(len(df)):
        for j in range(i + 1, len(df)):
          if isinstance(column_name, str):
            similarity = calculate_similarity(str(df.iloc[i, df_updated.columns.get_loc(column_name)]), str(df.iloc[j, df_updated.columns.get_loc(column_name)]))
          elif isinstance(column_name, list):
            similarity = mean([calculate_similarity(str(df.iloc[i, df_updated.columns.get_loc(col)]), str(df.iloc[j, df_updated.columns.get_loc(col)])) for col in column_name])
          else :
            raise ValueError("column_name doit être une chaîne de caractères ou une liste de chaînes de caractères")
          if similarity >= threshold:
              duplicates.append((i, j))
    return duplicates

#Identifie les index des colonnes pour les manip a venir
irating = df_updated.columns.get_loc('rating')
inumber_of_reviews = df_updated.columns.get_loc('number_of_reviews')

##On utilise la structure de la fonction gestion_doublons pour implementer la gestion de la note demande
def gestion_doublons_ratings(dataframe, list_doublons) :
  ##Gestion des doublons idententifier dans la liste
  for (i,j) in list_doublons:
    #on n'a pas a gerer d'ou vient les valeurs, tant que l'entrer i dans la fonction d'identification est l'entrer i ou og ici
    og = i
    supp = j

    #remplacer les NaN par des 0 pour nos manip
    if pd.isna(dataframe.iloc[og, irating]):
      og_rating = 0
    else :
      og_rating = dataframe.iloc[og, irating]
    if pd.isna(dataframe.iloc[supp, irating]):
      supp_rating = 0
    else :
      supp_rating = dataframe.iloc[supp, irating]
    if pd.isna(dataframe.iloc[og, inumber_of_reviews]):
      og_number_of_reviews = 0
    else :
      og_number_of_reviews = dataframe.iloc[og, inumber_of_reviews]
    if pd.isna(dataframe.iloc[supp, inumber_of_reviews]):
      supp_number_of_reviews = 0
    else :
      supp_number_of_reviews = dataframe.iloc[supp, inumber_of_reviews]

    #remplacer les valeurs du premier tuple (aka og aka i) puisque c'est celui qu'on va garder, les valeurs dans j deviennent non-importante
    dataframe.iloc[og, irating] = (og_rating*og_number_of_reviews + supp_rating*supp_number_of_reviews)/(og_number_of_reviews + supp_number_of_reviews)
    dataframe.iloc[og, inumber_of_reviews] = og_number_of_reviews + supp_number_of_reviews

  #drop les doublons avec des upc identiques, juste garder le premier
  dataframe.drop_duplicates(subset=['upc'], keep='first', inplace=True)

  #retourne le dataframe resultant de toutes ces operations
  return dataframe

dub_upc=identify_semantic_duplicates(df_updated, 'upc', 100)
updated_df=gestion_doublons_ratings(df_updated, dub_upc)
print(updated_df.head())
```
